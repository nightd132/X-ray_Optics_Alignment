```python
from pathlib import Path
PROJECT_ROOT = Path("..")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
experiment_summary_path = PROJECT_ROOT / "outputs/checkpoints/experiments_summary.csv"
exp_df = pd.read_csv(experiment_summary_path)
exp_df = exp_df.sort_values(by=["learning_rate"])
exp_df
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    exp_df["learning_rate"],
    exp_df["best_val_loss"],
    marker="o"
)

plt.xscale("log")
plt.xlabel("Learning Rate")
plt.ylabel("Best val loss")
plt.title("Effect of learning rate to val loss")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    exp_df["learning_rate"].astype(str),
    exp_df["best_epoch"]
)

plt.xlabel("Learning Rate")
plt.ylabel("Best Epoch")
plt.title("Best Epoch vs Learning Rate")
plt.show()
```

```python
path = (
    PROJECT_ROOT
        / "outputs/checkpoints/binsize_12_epoch_150_lr_0.01_batch_64_weightdecay_0.01"
        / "test_predictions.csv"
)
df = pd.read_csv(path)

fig, ax = plt.subplots(1, 3, figsize=(15, 4.5))
for k, n in enumerate(["x", "z"]):
    ax[k].scatter(df[f"target_{n}"], df[f"pred_{n}"], s=2, alpha=0.3)
    lim = [df[f"target_{n}"].min(), df[f"target_{n}"].max()]
    ax[k].plot(lim, lim, "r--")
    rmse = np.sqrt((df[f"err_{n}"] ** 2).mean())
    ax[k].set(xlabel=f"target {n}", ylabel=f"pred {n}", title=f"{n.upper()}: RMSE {rmse:.4g}")
ax[2].hist(df["err_x"], bins=100, alpha=0.6, label="X")
ax[2].hist(df["err_z"], bins=100, alpha=0.6, label="Z")
ax[2].set(title="Error distribution"); ax[2].legend()
plt.tight_layout(); plt.savefig(PROJECT_ROOT / "outputs/checkpoints/binsize_12_epoch_150_lr_0.01_batch_64_weightdecay_0.01"/"test_predictions.png", dpi=150)
```
